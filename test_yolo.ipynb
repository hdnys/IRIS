```python
from ultralytics import YOLO
model = YOLO('yolo12n.pt')
```

## Classes détectées par le modèle (Filtrage possible)

YOLO12 est entraîné sur COCO (80 classes). Liste-les pour choisir lesquelles garder.

```python
for i, name in model.names.items():
    print(f"{i:3d}  {name}")
```

### Filtrer

Passe `classes=[ids]` à `model(...)` pour ne garder que celles-là.
Ex: `classes=[0, 2, 56]` → personnes, voitures, chaises uniquement.

```python
KEEP = [0, 2, 56]  # à modifier selon le besoin

results = model('https://ultralytics.com/images/bus.jpg', classes=KEEP)
results[0].show()
for box in results[0].boxes:
    print(f"{model.names[int(box.cls[0])]}: {float(box.conf[0]):.2f}")
```

## Run on your webcam (press q to quit)

```python
import cv2

cap = cv2.VideoCapture(0)
while True:
    ok, frame = cap.read()
    if not ok:
        break
    annotated = model(frame)[0].plot()
    cv2.imshow('YOLO', annotated)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
cap.release()
cv2.destroyAllWindows()
```
